# Análisis exploratorio de datos

Contesta las siguientes preguntas escribiendo el código de Python necesario para encontrar las respuestas o producir las grficas necesarias

---


```python
# No mostrar advertencias
import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

# Cargar los datos
df = pd.read_csv('data/retailmax.csv')

# Configurar estilo
sns.set_style('whitegrid')
print('Datos cargados exitosamente!')
print(f'Dimensiones del dataset: {df.shape}')
print(df.head())
```

### 1. ¿Cual es la distribución de edades de los clientes?
- **Objetivo**: Comprender la distribución de las edades en el conjunto de datos.
- **Respuesta esperada**: Un histograma de la columna Age.

```python
# Histograma de edades en color rosa
plt.figure(figsize=(10, 6))
plt.hist(df['Age'], bins=20, color='pink', edgecolor='black', linewidth=1.2)
plt.title('Distribucion de Edades de los Clientes', fontsize=14, fontweight='bold')
plt.xlabel('Edad', fontsize=12)
plt.ylabel('Cantidad de Clientes', fontsize=12)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

print(f'Edad minima: {df["Age"].min()}')
print(f'Edad maxima: {df["Age"].max()}')
print(f'Edad promedio: {df["Age"].mean():.2f}')
```

### 2. ¿Existen diferencias significativas en los ingresos anuales entre hombres y mujeres?
- **Objetivo**: Comparar la distribución de ingresos anuales entre géneros.
- **Respuesta esperada**: Una grafica de cajas (boxplot) que muestre la distribución de Annual Income (k$) por género.

```python
# Boxplot de ingresos por genero en color rosa
plt.figure(figsize=(10, 6))
box = plt.boxplot([df[df['Gender'] == 'Male']['Annual Income (k$)'], 
                   df[df['Gender'] == 'Female']['Annual Income (k$)']],
                  labels=['Male', 'Female'],
                  patch_artist=True,
                  widths=0.5)

# Colorear las cajas de rosa
for patch in box['boxes']:
    patch.set_facecolor('pink')
    patch.set_edgecolor('black')
    patch.set_linewidth(1.5)

for median in box['medians']:
    median.set_color('black')
    median.set_linewidth(2)

plt.title('Distribucion de Ingresos Anuales por Genero', fontsize=14, fontweight='bold')
plt.xlabel('Genero', fontsize=12)
plt.ylabel('Ingreso Anual (k$)', fontsize=12)
plt.tight_layout()
plt.show()

print('Estadisticas por genero:')
print(df.groupby('Gender')['Annual Income (k$)'].describe())
```

### 3. ¿Como se distribuye la puntuacion de gasto (Spending Score) entre los diferentes rangos de edad?
- **Objetivo**: Analizar la relacion entre la edad y la puntuacion de gasto.
- **Respuesta esperada**: Una grafica de dispersion (scatter plot) o un grafico de cajas que compare la Spending Score (1-100) con diferentes grupos de edad.

```python
# Scatter plot de Spending Score vs Edad en color rosa
plt.figure(figsize=(10, 6))
plt.scatter(df['Age'], df['Spending Score (1-100)'], 
            color='pink', edgecolors='black', alpha=0.7, s=80, linewidth=0.8)
plt.title('Puntuacion de Gasto vs Edad', fontsize=14, fontweight='bold')
plt.xlabel('Edad', fontsize=12)
plt.ylabel('Spending Score (1-100)', fontsize=12)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

print('No hay una relacion clara entre edad y spending score basado en la dispersion de puntos.')
```

### 4. ¿Cual es la correlacion entre el ingreso anual y la puntuacion de gasto?
- **Objetivo**: Identificar si existe una relacion lineal entre el ingreso y el gasto.
- **Respuesta esperada**: Una grafica de dispersion y el calculo del coeficiente de correlacion entre Annual Income (k$) y Spending Score (1-100).

```python
# Grafica de correlacion entre Ingreso y Spending Score
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Scatter plot
ax1.scatter(df['Annual Income (k$)'], df['Spending Score (1-100)'], 
            color='pink', edgecolors='black', alpha=0.7, s=80, linewidth=0.8)
ax1.set_title('Relacion entre Ingreso Anual y Puntuacion de Gasto', fontsize=14, fontweight='bold')
ax1.set_xlabel('Ingreso Anual (k$)', fontsize=12)
ax1.set_ylabel('Spending Score (1-100)', fontsize=12)
ax1.grid(alpha=0.3)

# Calcular correlacion
correlacion = df['Annual Income (k$)'].corr(df['Spending Score (1-100)'])

# Mostrar correlacion en el grafico
ax1.text(0.05, 0.95, f'Coeficiente de Correlacion: {correlacion:.3f}', 
         transform=ax1.transAxes, fontsize=12, fontweight='bold',
         bbox=dict(boxstyle='round', facecolor='pink', alpha=0.8))

# Heatmap de correlacion
from matplotlib.colors import LinearSegmentedColormap
colors = ['#FFE6EA', '#FFB6C1', '#FF69B4']
pink_cmap = LinearSegmentedColormap.from_list('pink_custom', colors, N=256)

corr_matrix = df[['Annual Income (k$)', 'Spending Score (1-100)']].corr()
sns.heatmap(corr_matrix, annot=True, cmap=pink_cmap, ax=ax2, 
            fmt='.3f', square=True, linewidths=2, 
            cbar_kws={'label': 'Coeficiente de Correlacion'})
ax2.set_title('Matriz de Correlacion', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()

print(f'Coeficiente de correlacion entre Ingreso Anual y Spending Score: {correlacion:.4f}')
print('Interpretacion: La correlacion es muy cercana a 0, lo que indica que no hay una relacion lineal significativa entre ingreso y spending score.')
```

### 5. ¿Como varia la puntuacion de gasto en diferentes grupos de ingresos anuales?
- **Objetivo**: Examinar como los clientes en diferentes rangos de ingresos se comportan en terminos de gasto.
- **Respuesta esperada**: Una grafica de cajas o de violin que muestre la Spending Score (1-100) para diferentes rangos de Annual Income (k$).

```python
# Crear rangos de ingreso
df['Income_Range'] = pd.cut(df['Annual Income (k$)'], 
                             bins=[0, 30, 60, 90, 120, 150],
                             labels=['0-30k', '30-60k', '60-90k', '90-120k', '120-150k'])

# Boxplot de Spending Score por rango de ingreso
plt.figure(figsize=(10, 6))
box = plt.boxplot([df[df['Income_Range'] == cat]['Spending Score (1-100)'].dropna() 
                   for cat in df['Income_Range'].cat.categories],
                  patch_artist=True,
                  widths=0.5)

# Colorear las cajas de rosa
for patch in box['boxes']:
    patch.set_facecolor('pink')
    patch.set_edgecolor('black')
    patch.set_linewidth(1.5)

for median in box['medians']:
    median.set_color('black')
    median.set_linewidth(2)

plt.title('Puntuacion de Gasto por Rango de Ingreso Anual', fontsize=14, fontweight='bold')
plt.xlabel('Rango de Ingreso Anual (k$)', fontsize=12)
plt.ylabel('Spending Score (1-100)', fontsize=12)
plt.xticks(range(1, 6), ['0-30k', '30-60k', '60-90k', '90-120k', '120-150k'])
plt.tight_layout()
plt.show()

print('Spending Score promedio por rango de ingreso:')
print(df.groupby('Income_Range')['Spending Score (1-100)'].mean())
```

### 6. ¿Cual es la proporcion de clientes por genero?
- **Objetivo**: Determinar el balance de genero en el conjunto de datos.
- **Respuesta esperada**: Una grafica de barras o un grafico de pastel que muestre la proporcion de hombres y mujeres.

```python
# Grafica de proporcion por genero
gender_counts = df['Gender'].value_counts()

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# Grafica de barras
colors_bars = ['pink', 'lightcoral']
bars = ax1.bar(gender_counts.index, gender_counts.values, color=colors_bars, edgecolor='black', linewidth=1.5)
ax1.set_title('Proporcion de Clientes por Genero (Barras)', fontsize=14, fontweight='bold')
ax1.set_xlabel('Genero', fontsize=12)
ax1.set_ylabel('Cantidad de Clientes', fontsize=12)
ax1.grid(axis='y', alpha=0.3)

# Anadir porcentajes en las barras
for i, (idx, val) in enumerate(gender_counts.items()):
    percentage = (val / len(df)) * 100
    ax1.text(i, val + 2, f'{val}\n({percentage:.1f}%)', ha='center', va='bottom', fontsize=11, fontweight='bold')

# Grafica de pastel
colors_pie = ['pink', 'lightcoral']
wedges, texts, autotexts = ax2.pie(gender_counts.values, labels=gender_counts.index, 
                                    autopct='%1.1f%%', colors=colors_pie,
                                    explode=(0.05, 0.05), startangle=90,
                                    wedgeprops=dict(edgecolor='black', linewidth=1.5))
for autotext in autotexts:
    autotext.set_color('black')
    autotext.set_fontsize(12)
    autotext.set_fontweight('bold')
for text in texts:
    text.set_fontsize(13)
    text.set_fontweight('bold')

ax2.set_title('Proporcion de Clientes por Genero (Pastel)', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()

print('Distribucion por genero:')
for idx, val in gender_counts.items():
    print(f'{idx}: {val} clientes ({(val/len(df))*100:.1f}%)')
```

### 7. ¿Que grupos de edad gastan mas en promedio?
- **Objetivo**: Identificar los grupos de edad que tienen una mayor puntuacion de gasto en promedio.
- **Respuesta esperada**: Una grafica de barras que compare la puntuacion de gasto promedio entre diferentes grupos de edad.

```python
# Crear grupos de edad
df['Age_Group'] = pd.cut(df['Age'], 
                          bins=[0, 25, 35, 45, 55, 100],
                          labels=['<25', '25-35', '35-45', '45-55', '>55'])

# Calcular spending score promedio por grupo de edad
avg_spending_by_age = df.groupby('Age_Group')['Spending Score (1-100)'].mean().reindex(['<25', '25-35', '35-45', '45-55', '>55'])

# Grafica de barras
plt.figure(figsize=(10, 6))
bars = plt.bar(avg_spending_by_age.index, avg_spending_by_age.values, 
               color='pink', edgecolor='black', linewidth=1.5)
plt.title('Puntuacion de Gasto Promedio por Grupo de Edad', fontsize=14, fontweight='bold')
plt.xlabel('Grupo de Edad', fontsize=12)
plt.ylabel('Spending Score Promedio (1-100)', fontsize=12)
plt.grid(axis='y', alpha=0.3)

# Anadir valores en las barras
for i, (idx, val) in enumerate(avg_spending_by_age.items()):
    plt.text(i, val + 0.5, f'{val:.1f}', ha='center', va='bottom', fontsize=11, fontweight='bold')

plt.tight_layout()
plt.show()

print('Spending Score promedio por grupo de edad:')
print(avg_spending_by_age)
print(f'\nGrupo que mas gasta: {avg_spending_by_age.idxmax()} con {avg_spending_by_age.max():.1f} puntos')
```

### 8. ¿Hay alguna relacion entre la edad y el ingreso anual de los clientes?
- **Objetivo**: Explorar si hay una tendencia entre la edad de los clientes y sus ingresos.
- **Respuesta esperada**: Una grafica de dispersion que muestre la relacion entre Age y Annual Income (k$).

```python
# Scatter plot de Edad vs Ingreso Anual
plt.figure(figsize=(10, 6))
plt.scatter(df['Age'], df['Annual Income (k$)'], 
            color='pink', edgecolors='black', alpha=0.7, s=80, linewidth=0.8)
plt.title('Relacion entre Edad e Ingreso Anual', fontsize=14, fontweight='bold')
plt.xlabel('Edad', fontsize=12)
plt.ylabel('Ingreso Anual (k$)', fontsize=12)
plt.grid(alpha=0.3)

# Calcular y mostrar correlacion
corr_age_income = df['Age'].corr(df['Annual Income (k$)'])
plt.text(0.05, 0.95, f'Correlacion: {corr_age_income:.3f}', 
         transform=plt.gca().transAxes, fontsize=12, fontweight='bold',
         bbox=dict(boxstyle='round', facecolor='pink', alpha=0.8))

plt.tight_layout()
plt.show()

print(f'Coeficiente de correlacion entre Edad e Ingreso Anual: {corr_age_income:.4f}')
print('Interpretacion: La correlacion es muy cercana a 0, indicando que no hay una relacion lineal significativa entre edad e ingreso.')
```

### 9. ¿Cual es la distribucion conjunta de la edad y el ingreso anual?
- **Objetivo**: Entender como se distribuyen estas dos variables en conjunto.
- **Respuesta esperada**: Una grafica de dispersion con una densidad de puntos o un grafico de hexagonos que muestre la distribucion conjunta de Age y Annual Income (k$).

```python
# Distribucion conjunta de Edad e Ingreso
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Hexbin plot
hb = ax1.hexbin(df['Age'], df['Annual Income (k$)'], bins=30, cmap='pink_r', 
                edgecolors='white', linewidth=0.3, mincnt=1)
ax1.set_title('Distribucion Conjunta: Edad vs Ingreso (Hexbin)', fontsize=14, fontweight='bold')
ax1.set_xlabel('Edad', fontsize=12)
ax1.set_ylabel('Ingreso Anual (k$)', fontsize=12)
ax1.grid(alpha=0.3)
plt.colorbar(hb, ax=ax1, label='Cantidad de Clientes')

# Scatter con densidad (KDE)
ax2.scatter(df['Age'], df['Annual Income (k$)'], 
            color='pink', edgecolors='black', alpha=0.5, s=60, linewidth=0.8)

# Anadir contorno de densidad
sns.kdeplot(x=df['Age'], y=df['Annual Income (k$)'], 
            ax=ax2, levels=10, color='black', linewidths=2, alpha=0.6)

ax2.set_title('Distribucion Conjunta: Edad vs Ingreso (KDE)', fontsize=14, fontweight='bold')
ax2.set_xlabel('Edad', fontsize=12)
ax2.set_ylabel('Ingreso Anual (k$)', fontsize=12)
ax2.grid(alpha=0.3)

plt.tight_layout()
plt.show()

print('La distribucion muestra que los clientes estan dispersos a lo largo de todas las edades y rangos de ingreso.')
print('No hay concentraciones extremas en ninguna region especifica del espacio edad-ingreso.')
```

### 10. ¿Como se distribuyen los clientes en funcion de la puntuacion de gasto y el genero?
- **Objetivo**: Analizar la relacion entre la puntuacion de gasto y el genero.
- **Respuesta esperada**: Una grafica de dispersion o un grafico de violin que muestre la Spending Score (1-100) separada por genero.

```python
# Violin plot de Spending Score por genero
plt.figure(figsize=(10, 6))

sns.violinplot(x='Gender', y='Spending Score (1-100)', data=df, 
               palette=['pink', 'lightcoral'],
               edgecolor='black', linewidth=1.5)

plt.title('Distribucion de Puntuacion de Gasto por Genero', fontsize=14, fontweight='bold')
plt.xlabel('Genero', fontsize=12)
plt.ylabel('Spending Score (1-100)', fontsize=12)
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

print('Estadisticas de Spending Score por genero:')
print(df.groupby('Gender')['Spending Score (1-100)'].describe())
print('\nInterpretacion: Ambos generos muestran distribuciones similares de spending score, con medianas cercanas (47-50).')
```
